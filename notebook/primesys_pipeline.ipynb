```python
# Primesys RAG pipeline

# Process all text files in dirs

import os
from pathlib import Path

from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter


# Read all the text file in dirs

def process_all_texts(text_directory):
    
    all_documents = []
    text_dir = Path(text_directory)
    
    text_files = list(text_dir.glob("**/*.txt"))
    
    print(f"Found {len(text_files)} Text files to process")
    
    for text_file in text_files:
        print(f"\n processing: {text_file.name}")
        
        try:
            
            loader = TextLoader(str(text_file))
            documents = loader.load()
            
            for doc in documents:
                doc.metadata['source_file'] = text_file.name
                doc.metadata['file_type'] = 'txt'
                
            all_documents.extend(documents)
            print(f"Loaded {len(documents)} pages")
            
        except Exception as e:
            print(f"X Error: {e}")
            raise
    
    return all_documents


all_text_docs = process_all_texts("../data")
```
